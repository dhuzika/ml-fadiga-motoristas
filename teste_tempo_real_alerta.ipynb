# Sistema de Detecção de Fadiga

Sistema que detecta fadiga usando webcam e modelo treinado.
Escolha entre modelo XGBoost normal ou otimizado com Optuna.

## Funcionalidades
- **Seleção automática de modelo**: Escolha entre modelo normal e Optuna
- **Calibração personalizada**: Sistema aprende padrões de cada usuário  
- **Detecção em tempo real**: Análise de EAR, MAR, PERCLOS e blink rate
- **Interface visual**: Métricas e status em tempo real
- **Perfis de usuário**: Configurações salvas automaticamente

## Como usar
1. Execute a célula de **Seleção do Modelo** 
2. Execute as células seguintes em ordem
3. Faça a calibração inicial (20-25s)
4. Use o sistema normalmente

## Controles durante uso
- **q**: Sair
- **r**: Recalibrar  
- **p**: Salvar perfil"

```python
# Seleção do Modelo
from pathlib import Path
import json

def selecionar_modelo():
    print("=== SELEÇÃO DO MODELO ===")
    print("Modelos disponíveis:")
    
    modelos_disponiveis = []
    
    # Verifica modelo normal
    modelo_normal_dir = Path("modelos_xgb")
    if modelo_normal_dir.exists() and (modelo_normal_dir / "modelo_xgb.joblib").exists():
        try:
            with open(modelo_normal_dir / "info_classes.json", 'r') as f:
                info_normal = json.load(f)
            accuracy_normal = info_normal.get('accuracy', 0.0)
            print(f"1. Modelo Normal - Acurácia: {accuracy_normal:.1%}")
            modelos_disponiveis.append(('normal', modelo_normal_dir, accuracy_normal))
        except:
            print("1. Modelo Normal - (erro ao ler info)")
            modelos_disponiveis.append(('normal', modelo_normal_dir, 0.0))
    else:
        print("1. Modelo Normal - Não encontrado")
    
    # Verifica modelo Optuna
    modelo_optuna_dir = Path("modelos_xgb_optuna")
    if modelo_optuna_dir.exists() and (modelo_optuna_dir / "modelo_xgb.joblib").exists():
        try:
            with open(modelo_optuna_dir / "info_classes.json", 'r') as f:
                info_optuna = json.load(f)
            accuracy_optuna = info_optuna.get('accuracy', 0.0)
            auc_optuna = info_optuna.get('auc', 0.0)
            n_trials = info_optuna.get('n_trials', 0)
            print(f"2. Modelo Optuna - Acurácia: {accuracy_optuna:.1%} | AUC: {auc_optuna:.3f} | Trials: {n_trials}")
            modelos_disponiveis.append(('optuna', modelo_optuna_dir, accuracy_optuna))
        except:
            print("2. Modelo Optuna - (erro ao ler info)")
            modelos_disponiveis.append(('optuna', modelo_optuna_dir, 0.0))
    else:
        print("2. Modelo Optuna - Não encontrado")
    
    if not modelos_disponiveis:
        print("Nenhum modelo encontrado!")
        print("Execute primeiro os notebooks de treino:")
        print("- treino_xgboost.ipynb (modelo normal)")
        print("- treino_xgboost_optuna.ipynb (modelo Optuna)")
        return None, None
    
    print("\nEscolha um modelo:")
    
    # Auto-seleção do melhor modelo ou permitir escolha manual
    if len(modelos_disponiveis) == 1:
        modelo_escolhido = modelos_disponiveis[0]
        print(f"Auto-selecionado: {modelo_escolhido[0].title()}")
    else:
        # Recomenda o melhor por acurácia
        melhor_modelo = max(modelos_disponiveis, key=lambda x: x[2])
        print(f"Recomendado: {melhor_modelo[0].title()} (melhor acurácia)")
        
        while True:
            escolha = input("\nEscolha (1=Normal, 2=Optuna, Enter=Recomendado): ").strip()
            
            if escolha == "" or escolha.lower() in ['r', 'rec', 'recomendado']:
                modelo_escolhido = melhor_modelo
                break
            elif escolha == "1" and any(m[0] == 'normal' for m in modelos_disponiveis):
                modelo_escolhido = next(m for m in modelos_disponiveis if m[0] == 'normal')
                break
            elif escolha == "2" and any(m[0] == 'optuna' for m in modelos_disponiveis):
                modelo_escolhido = next(m for m in modelos_disponiveis if m[0] == 'optuna')
                break
            else:
                print("Opção inválida. Tente novamente.")
    
    tipo_modelo, diretorio_modelo, accuracy = modelo_escolhido
    print(f"\nModelo selecionado: {tipo_modelo.title()}")
    print(f"Diretório: {diretorio_modelo}")
    print(f"Acurácia: {accuracy:.1%}")
    
    return tipo_modelo, str(diretorio_modelo)

# Executa seleção
tipo_modelo_selecionado, modelo_dir_selecionado = selecionar_modelo()

if modelo_dir_selecionado:
    print(f"\nModelo {tipo_modelo_selecionado.title()} pronto para uso!")
else:
    print("Erro na seleção do modelo. Execute as células de treino primeiro.")
```

```python
# Detector de Fadiga com Calibração Personalizada
# Sistema que aprende os padrões de cada usuário

import json
from pathlib import Path

class DetectorFadiga:
    def __init__(self, pipeline, usuario="default"):
        self.pipeline = pipeline
        self.usuario = usuario
        
        # Buffers para dados
        self.buffer = deque(maxlen=90)
        self.ear_history = deque(maxlen=200)
        self.mar_history = deque(maxlen=200)
        self.piscadas = []
        
        # Pontos dos olhos e boca
        self.olho_direito = [33, 160, 158, 133, 153, 144]
        self.olho_esquerdo = [362, 385, 387, 263, 373, 380]
        self.boca = [13, 14, 78, 308, 0, 17]
        
        # Configurações do usuário
        self.config = {
            'ear_normal': 0.30,
            'ear_limite': 0.25,
            'mar_normal': 0.60,
            'blink_normal': 18.0,
            'calibrado': False,
            'sessoes': 0
        }
        
        # Estados atuais
        self.ear_atual = 0.0
        self.mar_atual = 0.0
        self.blink_rate = 0.0
        self.predicao = "Alerta"
        self.confianca = 0.0
        
        self.carregar_config()
    
    def salvar_config(self):
        try:
            pasta = Path("perfis_usuarios")
            pasta.mkdir(exist_ok=True)
            
            arquivo = pasta / f"{self.usuario}.json"
            with open(arquivo, 'w') as f:
                json.dump(self.config, f, indent=2)
        except Exception as e:
            print(f"Erro ao salvar: {e}")
    
    def carregar_config(self):
        try:
            arquivo = Path("perfis_usuarios") / f"{self.usuario}.json"
            if arquivo.exists():
                with open(arquivo, 'r') as f:
                    config_salva = json.load(f)
                    self.config.update(config_salva)
                print(f"Perfil carregado: {self.config['sessoes']} sessões")
        except Exception as e:
            print(f"Erro ao carregar: {e}")
    
    def calibrar(self, cap, tempo=20):
        print(f"Calibração ({tempo}s)")
        print("Fique natural e pisque normalmente")
        
        ear_dados = []
        mar_dados = []
        piscadas_cal = []
        
        inicio = time.time()
        frames = 0
        ear_historico = []
        
        while time.time() - inicio < tempo:
            ret, frame = cap.read()
            if not ret:
                continue
                
            frame = cv2.flip(frame, 1)
            h, w = frame.shape[:2]
            
            rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            results = face_mesh.process(rgb_frame)
            
            if results.multi_face_landmarks:
                landmarks = results.multi_face_landmarks[0].landmark
                
                ear_d = self.calcular_ear(self.olho_direito, landmarks, w, h)
                ear_e = self.calcular_ear(self.olho_esquerdo, landmarks, w, h)
                ear = (ear_d + ear_e) / 2.0
                
                mar = self.calcular_mar(landmarks, w, h)
                
                if ear > 0.1 and mar > 0:
                    ear_dados.append(ear)
                    mar_dados.append(mar)
                    ear_historico.append(ear)
                    
                    # Detectar piscadas durante calibração
                    if len(ear_historico) >= 10:
                        tendencia = np.diff(ear_historico[-10:])
                        
                        for i in range(len(tendencia) - 3):
                            if (tendencia[i] < -0.04 and 
                                tendencia[i+1] < 0 and 
                                tendencia[i+2] > 0.04):
                                
                                min_ear = min(ear_historico[i:i+3])
                                if min_ear < 0.3:
                                    piscadas_cal.append(min_ear)
            
            # Interface de calibração
            progresso = (time.time() - inicio) / tempo
            bar_w = int(w * 0.7)
            bar_x = (w - bar_w) // 2
            bar_y = h // 2
            
            cv2.rectangle(frame, (bar_x, bar_y), (bar_x + bar_w, bar_y + 25), (40, 40, 40), -1)
            cv2.rectangle(frame, (bar_x, bar_y), (bar_x + int(bar_w * progresso), bar_y + 25), (0, 200, 0), -1)
            
            tempo_restante = tempo - (time.time() - inicio)
            cv2.putText(frame, f"Calibracao: {tempo_restante:.0f}s", 
                       (bar_x, bar_y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)
            
            if len(ear_dados) > 0:
                cv2.putText(frame, f"EAR: {ear_dados[-1]:.3f}", (20, 40), 
                           cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)
                cv2.putText(frame, f"Piscadas: {len(piscadas_cal)}", (20, 70), 
                           cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)
            
            cv2.imshow('Calibracao', frame)
            
            if cv2.waitKey(1) & 0xFF == 27:
                return False
            
            frames += 1
        
        print(f"Dados: {len(ear_dados)} EAR, {len(piscadas_cal)} piscadas")
        
        if len(ear_dados) < 20:
            print("Poucas amostras")
            return False
        
        # Processar dados
        ear_limpo = self.limpar_dados(ear_dados)
        mar_limpo = self.limpar_dados(mar_dados)
        
        if len(ear_limpo) < 10:
            print("Dados com muito ruído")
            return False
        
        # Calcular configurações personalizadas
        ear_normal = np.percentile(ear_limpo, 60)
        mar_normal = np.percentile(mar_limpo, 50)
        ear_std = np.std(ear_limpo)
        
        self.config['ear_normal'] = ear_normal
        self.config['ear_limite'] = max(0.15, ear_normal - (2.0 * ear_std))
        self.config['mar_normal'] = mar_normal
        
        # Blink rate
        tempo_total = frames / 30.0
        if len(piscadas_cal) > 0:
            blink_baseline = (len(piscadas_cal) / tempo_total) * 60.0
        else:
            blink_baseline = 15.0
        
        self.config['blink_normal'] = max(8.0, min(35.0, blink_baseline))
        self.config['calibrado'] = True
        self.config['sessoes'] += 1
        
        self.salvar_config()
        
        print(f"Calibração concluída!")
        print(f"EAR: {ear_normal:.3f} (limite: {self.config['ear_limite']:.3f})")
        print(f"Blink rate: {blink_baseline:.1f} bpm")
        
        cv2.destroyWindow('Calibracao')
        return True
    
    def limpar_dados(self, dados):
        dados = np.array(dados)
        q1, q3 = np.percentile(dados, [25, 75])
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        return dados[(dados >= lower) & (dados <= upper)]
    
    def distancia(self, p1, p2):
        return np.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)
    
    def calcular_ear(self, pontos_olho, landmarks, w, h):
        try:
            pontos = []
            for idx in pontos_olho:
                x = landmarks[idx].x * w
                y = landmarks[idx].y * h
                pontos.append([x, y])
            
            if len(pontos) >= 6:
                d1 = self.distancia(pontos[1], pontos[5])
                d2 = self.distancia(pontos[2], pontos[4])
                d3 = self.distancia(pontos[0], pontos[3])
                
                if d3 > 0:
                    ear = (d1 + d2) / (2.0 * d3)
                    return max(0.0, min(1.0, ear))
            
            return 0.0
        except:
            return 0.0
    
    def calcular_mar(self, landmarks, w, h):
        try:
            pontos = []
            for idx in self.boca:
                x = landmarks[idx].x * w
                y = landmarks[idx].y * h
                pontos.append([x, y])
            
            if len(pontos) >= 6:
                d_vert1 = self.distancia(pontos[0], pontos[1])
                d_vert2 = self.distancia(pontos[4], pontos[5])
                d_horiz = self.distancia(pontos[2], pontos[3])
                
                if d_horiz > 0:
                    mar = (d_vert1 + d_vert2 * 0.5) / d_horiz
                    return max(0.0, min(3.0, mar))
            
            return 0.0
        except:
            return 0.0
    
    def detectar_piscada(self, ear):
        if len(self.ear_history) >= 10:
            historico = list(self.ear_history)[-10:]
            historico.append(ear)
            
            tendencia = np.diff(historico)
            limite_flex = self.config['ear_limite'] * 1.3
            
            for i in range(len(tendencia) - 4):
                queda = tendencia[i] < -0.025
                baixo = tendencia[i+1] < 0.015
                subida = tendencia[i+2] > 0.015 or tendencia[i+3] > 0.015
                
                if queda and subida:
                    min_ear = min(historico[i:i+4])
                    max_ear = max(historico[i:i+4])
                    
                    if (min_ear < limite_flex and 
                        max_ear > min_ear + 0.03 and 
                        max_ear < 0.6):
                        
                        agora = time.time()
                        if not self.piscadas or (agora - self.piscadas[-1]['tempo']) > 0.15:
                            self.piscadas.append({'tempo': agora, 'ear_min': min_ear})
                            return True
        
        return False
    
    def calcular_blink_rate(self):
        agora = time.time()
        janela = 30.0
        
        piscadas_recentes = [p for p in self.piscadas 
                           if agora - p['tempo'] <= janela]
        
        if len(piscadas_recentes) >= 2:
            tempo_real = min(janela, agora - self.piscadas[0]['tempo'])
            rate = (len(piscadas_recentes) / tempo_real) * 60.0
            return min(60.0, max(0.0, rate))
        
        return 0.0
    
    def processar_features(self, landmarks, w, h):
        ear_d = self.calcular_ear(self.olho_direito, landmarks, w, h)
        ear_e = self.calcular_ear(self.olho_esquerdo, landmarks, w, h)
        ear = (ear_d + ear_e) / 2.0
        
        mar = self.calcular_mar(landmarks, w, h)
        
        self.ear_history.append(ear)
        self.mar_history.append(mar)
        
        self.detectar_piscada(ear)
        
        # PERCLOS
        if len(self.ear_history) >= 60:
            window = list(self.ear_history)[-60:]
            frames_fechados = sum(1 for e in window if e < self.config['ear_limite'])
            perclos = (frames_fechados / len(window)) * 100
        else:
            perclos = 0.0
        
        blink_rate = self.calcular_blink_rate()
        
        # Head stability
        try:
            nose1 = landmarks[1]
            nose2 = landmarks[9]
            head_stability = abs(nose1.x - nose2.x) + abs(nose1.y - nose2.y)
        except:
            head_stability = 0.0
        
        self.ear_atual = ear
        self.mar_atual = mar
        self.blink_rate = blink_rate
        
        return [perclos, mar, blink_rate, head_stability]
    
    def processar_frame(self, frame):
        h, w = frame.shape[:2]
        rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        results = face_mesh.process(rgb_frame)
        
        if results.multi_face_landmarks:
            landmarks = results.multi_face_landmarks[0].landmark
            features = self.processar_features(landmarks, w, h)
            self.buffer.append(features)
            return True
        return False
    
    def predizer(self):
        if len(self.buffer) == 90:
            sequence = np.array(list(self.buffer))
            pred, probs, classe = self.pipeline.predict_sequence(sequence)
            
            self.predicao = classe
            self.confianca = max(probs.values())
            
            return True
        return False
    
    def desenhar_interface(self, frame):
        h, w = frame.shape[:2]
        
        # Painel principal
        overlay = frame.copy()
        cv2.rectangle(overlay, (10, 10), (550, 300), (0, 0, 0), -1)
        cv2.addWeighted(overlay, 0.7, frame, 0.3, 0, frame)
        
        # Status
        cor = (0, 255, 0) if self.predicao == "Alerta" else (0, 0, 255)
        cv2.putText(frame, f"Status: {self.predicao}", (20, 40), 
                   cv2.FONT_HERSHEY_SIMPLEX, 1.0, cor, 2)
        
        cv2.putText(frame, f"Confianca: {self.confianca:.1%}", (20, 75), 
                   cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)
        
        cv2.putText(frame, f"Buffer: {len(self.buffer)}/90", (20, 105), 
                   cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)
        
        # Perfil
        if self.config['calibrado']:
            cv2.putText(frame, f"Perfil: Ativo (sessao {self.config['sessoes']})", 
                       (20, 135), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 1)
        else:
            cv2.putText(frame, "Perfil: Nao calibrado", (20, 135), 
                       cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 0, 255), 1)
        
        # Features
        cv2.putText(frame, "=== METRICAS ===", (300, 40), 
                   cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 0), 1)
        
        cv2.putText(frame, f"EAR: {self.ear_atual:.3f}", (300, 70), 
                   cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 1)
        cv2.putText(frame, f"MAR: {self.mar_atual:.3f}", (300, 95), 
                   cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 1)
        
        # Blink rate com cor
        cor_blink = (0, 255, 0) if self.blink_rate > 0 else (0, 0, 255)
        cv2.putText(frame, f"Blink Rate: {self.blink_rate:.1f}", (300, 120), 
                   cv2.FONT_HERSHEY_SIMPLEX, 0.5, cor_blink, 1)
        
        cv2.putText(frame, f"Piscadas: {len(self.piscadas)}", (300, 145), 
                   cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 1)
        
        # Barra EAR
        cv2.putText(frame, "EAR vs Limite:", (20, 180), 
                   cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 1)
        
        bar_w = 200
        bar_x = 20
        bar_y = 190
        
        cv2.rectangle(frame, (bar_x, bar_y), (bar_x + bar_w, bar_y + 15), (50, 50, 50), -1)
        
        # Limite
        limite_pos = int((self.config['ear_limite'] / 0.5) * bar_w)
        cv2.line(frame, (bar_x + limite_pos, bar_y), (bar_x + limite_pos, bar_y + 15), (0, 255, 255), 2)
        
        # EAR atual
        ear_pos = int((self.ear_atual / 0.5) * bar_w)
        ear_pos = min(bar_w, max(0, ear_pos))
        cor_ear = (0, 255, 0) if self.ear_atual > self.config['ear_limite'] else (0, 0, 255)
        cv2.circle(frame, (bar_x + ear_pos, bar_y + 7), 5, cor_ear, -1)
        
        return frame

# Criar detector
detector = DetectorFadiga(pipeline, usuario="usuario1")
print("Detector criado!")
print(f"Status: {'Calibrado' if detector.config['calibrado'] else 'Novo usuário'}")
if detector.config['calibrado']:
    print(f"EAR limite: {detector.config['ear_limite']:.3f}")
    print(f"Sessões: {detector.config['sessoes']}")
```

```python
# Sistema de Detecção MENOS SENSÍVEL - Versão Melhorada
import cv2
import platform
import os
import time
from collections import deque

# Sistema de Suavização Temporal
class SuavizadorTemporal:
    def __init__(self, janela=6, threshold_consenso=0.67):
        """
        janela: número de predições para analisar
        threshold_consenso: % de predições que devem ser "Sonolento" para alertar
        """
        self.janela = janela
        self.threshold_consenso = threshold_consenso
        self.historico = deque(maxlen=janela)
        self.perclos_historico = deque(maxlen=janela)
    
    def adicionar_predicao(self, predicao, confianca, perclos):
        """Adiciona nova predição e PERCLOS ao histórico"""
        self.historico.append({
            'predicao': predicao,
            'confianca': confianca,
            'tempo': time.time()
        })
        self.perclos_historico.append(perclos)
        
        return self.deve_alertar()
    
    def deve_alertar(self):
        """Decide se deve alertar baseado no consenso temporal"""
        if len(self.historico) < 3:  # Precisa de pelo menos 3 predições
            return False
        
        # 1. Verifica PERCLOS médio (pré-condição)
        perclos_medio = sum(self.perclos_historico) / len(self.perclos_historico)
        if perclos_medio < 40.0:  # PERCLOS < 40% = não está com olhos fechados suficiente
            return False
        
        # 2. Conta predições "Sonolento" com alta confiança
        sonolento_alta_conf = 0
        for entrada in self.historico:
            if (entrada['predicao'] == "Sonolento" and 
                entrada['confianca'] > 0.95):  # Threshold mais alto!
                sonolento_alta_conf += 1
        
        # 3. Calcula percentual de consenso
        consenso = sonolento_alta_conf / len(self.historico)
        
        # 4. Retorna True apenas se threshold de consenso atingido
        return consenso >= self.threshold_consenso
    
    def get_info(self):
        """Retorna informações para debug"""
        if len(self.historico) == 0:
            return "Suavizador: Sem dados"
        
        sonolento_count = sum(1 for h in self.historico 
                             if h['predicao'] == "Sonolento" and h['confianca'] > 0.95)
        consenso = sonolento_count / len(self.historico) if self.historico else 0
        perclos_medio = sum(self.perclos_historico) / len(self.perclos_historico) if self.perclos_historico else 0
        
        return f"Consenso: {consenso:.1%} | PERCLOS: {perclos_medio:.1f}% | Buffer: {len(self.historico)}/{self.janela}"

def emitir_alerta_sonoro():
    """Emite um alerta sonoro simples (Linux)."""
    print("\nFADIGA DETECTADA - CONSENSO CONFIRMADO!")
    
    # Terminal bell
    for _ in range(3):
        print('\a', end='', flush=True)
    
    # Tentativa simples de beep (não-bloqueante)
    try:
        os.system('aplay /usr/share/sounds/alsa/Front_Left.wav >/dev/null 2>&1 &')
    except:
        pass

# Função principal com melhorias
def testar_sistema_melhorado():
    cv2.destroyAllWindows()
    cv2.waitKey(1)

    # Inicializa suavizador temporal
    suavizador = SuavizadorTemporal(janela=6, threshold_consenso=0.67)  # 67% de consenso
    
    print("SISTEMA MELHORADO INICIADO")
    print("Configurações:")
    print("- Threshold confiança: 95% (era 90%)")
    print("- Suavização temporal: 6 predições, consenso 67%")
    print("- Intervalo entre alertas: 10s (era 3s)")
    print("- Pré-condição: PERCLOS > 40%")
    print("-" * 50)

    # Tenta inicializar a câmera
    cap = cv2.VideoCapture(0)
    if not cap.isOpened():
        print("Erro: Não foi possível abrir a câmera.")
        print("Verifique se a câmera está conectada e não está sendo usada por outro aplicativo.")
        return

    cap.set(cv2.CAP_PROP_FRAME_WIDTH, 640)
    cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 480)
    cap.set(cv2.CAP_PROP_FPS, 30)

    print("Câmera inicializada")

    try:
        # Calibração
        if detector.config['calibrado']:
            print(f"Perfil encontrado (sessão {detector.config['sessoes']})")
            print(f"EAR limite: {detector.config['ear_limite']:.3f}")

            resposta = input("Recalibrar? (s/N): ").lower()
            if resposta in ['s', 'sim']:
                sucesso = detector.calibrar(cap, 20)
            else:
                sucesso = True
                print("Usando perfil existente")
        else:
            print("Novo usuário - fazendo calibração")
            sucesso = detector.calibrar(cap, 25)

        if not sucesso:
            print("Falha na calibração. Encerrando.")
            return

        # Detecção
        print("\nIniciando detecção MENOS SENSÍVEL")
        print("Controles: q=Sair, r=Recalibrar, p=Salvar")

        cv2.namedWindow('Detecção de Fadiga - Menos Sensível', cv2.WINDOW_NORMAL)
        cv2.resizeWindow('Detecção de Fadiga - Menos Sensível', 900, 650)

        contador = 0
        inicio = time.time()
        tempo_ultimo_alerta = 0
        perclos_atual = 0.0

        while True:
            ret, frame = cap.read()
            if not ret:
                print("Erro ao ler frame da câmera.")
                break

            frame = cv2.flip(frame, 1)
            h, w = frame.shape[:2]

            # Processar frame
            face_ok = detector.processar_frame(frame)
            predicao_ok = detector.predizer()

            # Calcular PERCLOS atual
            if len(detector.ear_history) >= 60:
                window = list(detector.ear_history)[-60:]
                frames_fechados = sum(1 for e in window if e < detector.config['ear_limite'])
                perclos_atual = (frames_fechados / len(window)) * 100

            # --- LÓGICA DE ALERTA MELHORADA ---
            if predicao_ok:
                # Adiciona ao suavizador temporal
                deve_alertar = suavizador.adicionar_predicao(
                    detector.predicao, 
                    detector.confianca, 
                    perclos_atual
                )
                
                # Alerta apenas se:
                # 1. Suavizador confirma consenso
                # 2. Passou tempo suficiente desde último alerta (10s)
                if deve_alertar:
                    agora = time.time()
                    if (agora - tempo_ultimo_alerta) > 10.0:  # 10 segundos!
                        emitir_alerta_sonoro()
                        tempo_ultimo_alerta = agora
            # ------------------------------------

            # Interface melhorada
            texto_cor = (255, 255, 255)
            titulo_cor = (0, 255, 255)
            alerta_cor = (0, 0, 255)
            ok_cor = (0, 255, 0)
            aviso_cor = (0, 165, 255)  # Laranja
            font_scale_status = 0.8
            font_scale_metricas = 0.6
            thickness = 2

            # Título do sistema
            cv2.putText(frame, "SISTEMA MENOS SENSIVEL", (10, 25), 
                       cv2.FONT_HERSHEY_SIMPLEX, 0.7, titulo_cor, 2)

            # Status principal
            status_text = f"Predicao: {detector.predicao}"
            status_color = alerta_cor if detector.predicao == "Sonolento" else ok_cor
            cv2.putText(frame, status_text, (10, 55), cv2.FONT_HERSHEY_SIMPLEX, font_scale_status, status_color, thickness)
            
            # Confiança com cor baseada no threshold
            conf_color = ok_cor if detector.confianca > 0.95 else aviso_cor
            cv2.putText(frame, f"Confianca: {detector.confianca:.1%}", (10, 85), 
                       cv2.FONT_HERSHEY_SIMPLEX, font_scale_metricas, conf_color, thickness)

            # Informações do suavizador
            suav_info = suavizador.get_info()
            cv2.putText(frame, suav_info, (10, 115), 
                       cv2.FONT_HERSHEY_SIMPLEX, 0.5, titulo_cor, 1)

            # PERCLOS atual
            perclos_color = alerta_cor if perclos_atual > 40 else ok_cor
            cv2.putText(frame, f"PERCLOS: {perclos_atual:.1f}%", (10, 145), 
                       cv2.FONT_HERSHEY_SIMPLEX, font_scale_metricas, perclos_color, thickness)

            # Métricas na direita
            metrics_x_pos = w - 220
            cv2.putText(frame, "METRICAS", (metrics_x_pos, 30), cv2.FONT_HERSHEY_SIMPLEX, font_scale_metricas, titulo_cor, thickness)
            cv2.putText(frame, f"EAR: {detector.ear_atual:.3f}", (metrics_x_pos, 60), cv2.FONT_HERSHEY_SIMPLEX, font_scale_metricas, texto_cor, thickness)
            cv2.putText(frame, f"MAR: {detector.mar_atual:.3f}", (metrics_x_pos, 85), cv2.FONT_HERSHEY_SIMPLEX, font_scale_metricas, texto_cor, thickness)
            cv2.putText(frame, f"Blink Rate: {detector.blink_rate:.1f}", (metrics_x_pos, 110), cv2.FONT_HERSHEY_SIMPLEX, font_scale_metricas, texto_cor, thickness)
            cv2.putText(frame, f"Piscadas: {len(detector.piscadas)}", (metrics_x_pos, 135), cv2.FONT_HERSHEY_SIMPLEX, font_scale_metricas, texto_cor, thickness)

            # Configurações melhoradas
            cv2.putText(frame, "CONFIGURACOES:", (metrics_x_pos, 170), cv2.FONT_HERSHEY_SIMPLEX, 0.5, titulo_cor, 1)
            cv2.putText(frame, "Threshold: 95%", (metrics_x_pos, 190), cv2.FONT_HERSHEY_SIMPLEX, 0.4, texto_cor, 1)
            cv2.putText(frame, "Consenso: 67%", (metrics_x_pos, 210), cv2.FONT_HERSHEY_SIMPLEX, 0.4, texto_cor, 1)
            cv2.putText(frame, "Intervalo: 10s", (metrics_x_pos, 230), cv2.FONT_HERSHEY_SIMPLEX, 0.4, texto_cor, 1)

            # Status da face
            status_face = "Face OK" if face_ok else "Sem face"
            cor_face = ok_cor if face_ok else alerta_cor
            cv2.putText(frame, status_face, (w - 120, h - 50), cv2.FONT_HERSHEY_SIMPLEX, 0.6, cor_face, 2)

            # Controles
            cv2.putText(frame, "q=Sair | r=Recalibrar | p=Salvar", (10, h - 20), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 1)

            cv2.imshow('Detecção de Fadiga - Menos Sensível', frame)

            # Estatísticas a cada 30 frames
            contador += 1
            if contador % 30 == 0:
                tempo_decorrido = time.time() - inicio
                if tempo_decorrido > 0:
                    fps = contador / tempo_decorrido
                    print(f"FPS: {fps:.1f} | Status: {detector.predicao} | "
                          f"Conf: {detector.confianca:.2f} | PERCLOS: {perclos_atual:.1f}% | {suav_info}")

            # Teclas
            key = cv2.waitKey(1) & 0xFF
            if key == ord('q') or key == 27:
                break
            elif key == ord('r'):
                print("\nRecalibrando...")
                detector.buffer.clear()
                detector.ear_history.clear()
                detector.piscadas.clear()
                suavizador = SuavizadorTemporal(janela=6, threshold_consenso=0.67)  # Reset suavizador
                cv2.destroyWindow('Detecção de Fadiga - Menos Sensível')
                if detector.calibrar(cap, 20):
                    print("Recalibração concluída")
                else:
                    print("Recalibração falhou")
                cv2.namedWindow('Detecção de Fadiga - Menos Sensível', cv2.WINDOW_NORMAL)
                cv2.resizeWindow('Detecção de Fadiga - Menos Sensível', 900, 650)
            elif key == ord('p'):
                detector.salvar_config()
                print("\nPerfil salvo!")

            if cv2.getWindowProperty('Detecção de Fadiga - Menos Sensível', cv2.WND_PROP_VISIBLE) < 1:
                break

    except KeyboardInterrupt:
        print("\nInterrompido pelo usuário")
    except Exception as e:
        print(f"\nOcorreu um erro inesperado: {e}")
    finally:
        # Libera a câmera e fecha as janelas
        cap.release()
        cv2.destroyAllWindows()
        cv2.waitKey(1)

        # Relatório final
        if 'inicio' in locals() and 'contador' in locals() and contador > 0:
            tempo_total = time.time() - inicio
            if tempo_total > 0:
                print(f"\n--- Relatório Final (Sistema Melhorado) ---")
                print(f"Tempo total de execução: {tempo_total:.1f}s")
                print(f"Total de frames processados: {contador}")
                print(f"FPS médio: {contador/tempo_total:.1f}")
                print(f"Total de piscadas detectadas: {len(detector.piscadas)}")
                if detector.config['calibrado']:
                    print(f"EAR limite utilizado: {detector.config['ear_limite']:.3f}")
                print("Configurações aplicadas:")
                print("- Threshold confiança: 95%")
                print("- Consenso temporal: 67% em 6 predições")
                print("- Intervalo alertas: 10 segundos")
                print("- Pré-condição PERCLOS: > 40%")

        detector.salvar_config()
        print("Sistema MENOS SENSÍVEL encerrado.")

# --- Execução do Sistema Melhorado ---
if 'detector' in globals():
    print("SISTEMA MENOS SENSÍVEL")
    testar_sistema_melhorado()
```
